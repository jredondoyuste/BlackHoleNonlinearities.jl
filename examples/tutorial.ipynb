# From homogeneous waves to $\mathcal Q$

Units: $G=c=M=1$.

```julia
using Plots
include(joinpath("..", "jl", "QFactor.jl"))
```

## Homogeneous solution

```julia
l, ω, rmax = 2, 0.23, 800.0
sol = linear_sol(l, ω, "even", 5.0, rmax)
r = range(2.01, 80.0, length=1200)
plot(r, real.(first.(sol.(r))), xlabel="r / M", ylabel="Re ψ⁽¹⁾", label=false)
```

```julia
ain, aout = extract_amps(sol, ω, rmax, "even")
(; ain, aout, reflection=abs2(aout / ain))
```

## Regularized source

```julia
sol2 = linear_sol(2, ω, "even", 5.0, rmax)
S = make_source("eee", sol, sol2, 2, 2, 2, 2, -2, ω, ω)
rfar = exp10.(range(log10(20.0), log10(rmax), length=200))
sfar = abs.(S.(rfar))
plot(rfar, sfar, xscale=:log10, yscale=:log10, xlabel="r / M", ylabel="|S|", label="source")
plot!(rfar, sfar[1] .* (rfar[1] ./ rfar).^2, linestyle=:dash, label="r⁻²")
```

## Coupling coefficient

```julia
mode1 = (2, 2, ω)
mode2 = (2, -2, ω)
result = qfactor_full("eee", mode1, mode2, 2; rmin=5.0, rmax=800.0)
(; result.Qin, result.Qout)
```

```julia
radii = [400.0, 800.0, 1600.0]
q = [qfactor_full("eee", mode1, mode2, 2; rmin=5.0, rmax=R).Qin for R in radii]
(; radii, Qin=q, relative_change=abs.(q .- q[end]) ./ abs(q[end]))
```
